```sql
SELECT
    usage_metadata.app_name,
    usage_metadata.app_id,
    identity_metadata.created_by as user, 
    sku_name,
    sum(usage_quantity) as total_usage
FROM system.billing.usage u
WHERE usage_metadata.app_name is not null
and usage_start_time>= trunc(current_timestamp(), 'month') - 30
group by usage_metadata.app_name, usage_metadata.app_id, identity_metadata.created_by, sku_name
```
