```sql
SELECT
    usage_date,
    usage_metadata.job_run_id AS job_run_id,
    sku_name,
    sum(usage_quantity) AS total_usage
FROM system.billing.usage
WHERE usage_date >= current_date() - INTERVAL 30 DAYS
    AND usage_date <  current_date() - INTERVAL 1 DAY
    AND usage_metadata.job_id IS NOT NULL
    AND sku_name = 'PREMIUM_JOBS_SERVERLESS_COMPUTE_US_EAST_2'
    and usage_metadata.job_id = '427409455473593'
GROUP BY usage_metadata.job_run_id, sku_name, usage_date
ORDER BY usage_date desc

```
